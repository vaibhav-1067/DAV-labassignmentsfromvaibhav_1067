```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("earthquake_cleaned.csv")

print(df.shape)
df.head()

tsunami_counts = df["tsunami"].value_counts().sort_index()

plt.bar(
    ["No Tsunami (0)", "Tsunami (1)"],
    [
        tsunami_counts.get(0, 0),
        tsunami_counts.get(1, 0)
    ]
)

plt.title("Earthquakes by Tsunami Occurrence")
plt.xlabel("Tsunami")
plt.ylabel("Number of Earthquakes")

plt.show()


plt.hist(df["magnitude"], bins=5)

plt.title("Distribution of Earthquake Magnitude - 5 Bins")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()

plt.hist(df["magnitude"], bins=10)

plt.title("Distribution of Earthquake Magnitude - 10 Bins")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()

plt.hist(df["magnitude"], bins=20)

plt.title("Distribution of Earthquake Magnitude - 20 Bins")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()

sns.boxplot(x=df["magnitude"])

plt.title("Box Plot of Earthquake Magnitude")
plt.xlabel("Magnitude")

plt.show()

sns.violinplot(x=df["magnitude"])

plt.title("Violin Plot of Earthquake Magnitude")
plt.xlabel("Magnitude")

plt.show()

tsunami_counts = df["tsunami"].value_counts().sort_index()

plt.pie(
    tsunami_counts,
    labels=["No Tsunami", "Tsunami"],
    autopct="%1.1f%%"
)

plt.title("Tsunami Occurrence")
plt.show()

plt.scatter(
    df["magnitude"],
    df["sig"],
    alpha=0.5
)

plt.title("Magnitude vs Significance")
plt.xlabel("Magnitude")
plt.ylabel("Significance")
plt.show()

df["date_time"] = pd.to_datetime(df["date_time"])

df["year"] = df["date_time"].dt.year

year_count = df.groupby("year").size()

plt.plot(
    year_count.index,
    year_count.values,
    marker="o"
)

plt.title("Earthquakes per Year")
plt.xlabel("Year")
plt.ylabel("Number of Earthquakes")

plt.show()


numeric = df.select_dtypes(include="number")

corr = numeric.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Earthquake Variables")
plt.show()

# now task 2 we have got all the 8 graphs
# honest  chart
import matplotlib.pyplot as plt

tsunami_counts = df["tsunami"].value_counts().sort_index()

plt.bar(
    ["No Tsunami (0)", "Tsunami (1)"],
    [tsunami_counts.get(0, 0), tsunami_counts.get(1, 0)]
)

plt.title("HONEST: Earthquakes by Tsunami Occurrence")
plt.xlabel("Tsunami")
plt.ylabel("Number of Earthquakes")
plt.ylim(0, 750)

plt.show()


# misleading chart

plt.bar(
    ["No Tsunami (0)", "Tsunami (1)"],
    [tsunami_counts.get(0, 0), tsunami_counts.get(1, 0)]
)

plt.title("MISLEADING: Earthquakes by Tsunami Occurrence")
plt.xlabel("Tsunami")
plt.ylabel("Number of Earthquakes")
plt.ylim(300, 700)

plt.show()

# pair 2 histogram bin manipultion
# honest graph
plt.hist(df["magnitude"], bins=10)

plt.title("HONEST: Earthquake Magnitude Distribution")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()

# misleading chart
plt.hist(df["magnitude"], bins=2)

plt.title("MISLEADING: Earthquake Magnitude Distribution")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()

# scatter plot : axis manipulation
#honest chart

plt.scatter(
    df["magnitude"],
    df["sig"],
    alpha=0.5
)

plt.title("HONEST: Magnitude vs Significance")
plt.xlabel("Magnitude")
plt.ylabel("Significance")
plt.xlim(6.4, 9.2)

plt.show()

# misleading chart

plt.scatter(
    df["magnitude"],
    df["sig"],
    alpha=0.5
)

plt.title("MISLEADING: Magnitude vs Significance")
plt.xlabel("Magnitude")
plt.ylabel("Significance")
plt.xlim(6.75, 7.1)

plt.show()
```
